```python
from langchain_core.messages import AIMessage

ai_msg = AIMessage(
    content = "Message from the AI"
)
print(ai_msg.text)
```

```python
from langchain_core.messages import HumanMessage

hm_msg = HumanMessage(
    content="Message from the human"
)
print(hm_msg.text)
```

```python
history = [ai_msg, hm_msg]
for m in history:
    m.pretty_print()
```

```python
history + [AIMessage(content="Message 2 form AI")]
```

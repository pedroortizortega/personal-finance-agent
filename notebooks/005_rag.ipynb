```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model="gpt-4o",
    temperature=0
)
response = llm.invoke("Hola como estás?")
response.pretty_print()
```

```python
# Id de base de datos vectorial
VectorStorage = [os.getenv("VECTOR_STORE_ID")]
```

```python
# declaracion para el usar el file search tool de OpenAI

file_search_tool = {
    "type": "file_search",
    "vector_store_ids": VectorStorage
}

llm_with_tools = llm.bind_tools([file_search_tool])
```

```python
response = llm_with_tools.invoke("Cómo mejoro el rendimiento de mi website?")
response.pretty_print()
```

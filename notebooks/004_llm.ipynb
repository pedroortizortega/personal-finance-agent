```python
from langchain_openai import ChatOpenAI

gpt4 = ChatOpenAI(
    model="gpt-4o",
    temperature=1
)
response = gpt4.invoke("Hola como estás?")
response.pretty_print()
```

```python
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage

################################
# history esta separado del invoke (del LLM)
msg_1 = SystemMessage("Eres un asistente de ayuda en español")
msg_2 = HumanMessage("me llammo Romeo")
msg_3 = AIMessage("Hola Romeo, cómo estás?")
msg_4 = HumanMessage("Cómo me llamo?")

history = [msg_1, msg_2, msg_3, msg_4]
################################

response = gpt4.invoke(history)
response.pretty_print()
```

```python
from langchain.chat_models import init_chat_model
# init_chat_model es por comodidad y solo configuras el nombre y no cargas la libreria especifica
llm = init_chat_model("gpt-4o", temperature=0.5)
response = llm.invoke(history)
response.pretty_print()
```

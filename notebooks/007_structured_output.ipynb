```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model

load_dotenv()
```

```python
llm = init_chat_model("gpt-4o", temperature=0)
response = llm.invoke("Holas me llaman Romeo y im telefono es 9425671235")
response.text
 
```

```python
from pydantic import BaseModel, Field

class ContactInfo(BaseModel):
    name: str = Field(..., description="The name of the person")
    email: str = Field(..., description="The email address of the person")
    phone: str = Field(..., description="The phone number of the person")
    tone: int = Field(..., description="The tone of the message, on a scale from 1 to 10")
    age: int = Field(..., description="The age of the person")
    sentiment: str = Field(..., description="The sentiment of the message, can be positive, negative, or neutral")

# Aqui basicamente le estamos diciendo al modelo que queremos que la salida sea estructurada y que siga el esquema definido en ContactInfo. Esto nos permite obtener una salida más organizada y fácil de manejar.
llm_with_structured_output = llm.with_structured_output(ContactInfo)

# aqui conviene agregar el system prompt para que el modelo sepa que queremos que siga el esquema definido en ContactInfo. Esto nos permite obtener una salida más organizada y fácil de manejar.
messages = [
    ("system", "You are a helpful assistant that extracts structured information from the conversation. Follow the ContactInfo schema."),
    ("user", "Hola me llaman Romeo y mi telefono es 9425671235. Mi correo es romeo@example.com")
]
response = llm_with_structured_output.invoke(messages)
response
```
